### 1. Risk Management Configuration & Paper Trading Simulator
Here we define a class to handle our paper trading portfolio state, track orders, and strictly enforce risk rules (like maximum daily drawdown, individual trade stop-loss, and target profit).

```python
class PaperTradingRiskManager:
    def __init__(self, initial_capital=100000.0, max_daily_loss=2000.0, default_stop_loss_pct=0.01, default_take_profit_pct=0.02):
        self.capital = initial_capital
        self.portfolio_value = initial_capital
        self.max_daily_loss = max_daily_loss
        self.default_stop_loss_pct = default_stop_loss_pct
        self.default_take_profit_pct = default_take_profit_pct

        self.positions = {}  # {stock_code: {'qty': qty, 'entry_price': price, 'sl': sl_price, 'tp': tp_price}}
        self.trade_log = []
        self.realized_pnl = 0.0
        self.active = True

    def execute_paper_order(self, stock_code, qty, current_price, order_type="BUY"):
        if not self.active:
            print("🚨 Risk Management Block: Trading is disabled due to risk limits reached.")
            return

        if order_type == "BUY":
            cost = qty * current_price
            if cost > self.capital:
                print(f"⚠️ Insufficient paper funds to buy {qty} shares of {stock_code}.")
                return

            self.capital -= cost
            sl_price = current_price * (1 - self.default_stop_loss_pct)
            tp_price = current_price * (1 + self.default_take_profit_pct)

            if stock_code in self.positions:
                # Average out existing position
                existing = self.positions[stock_code]
                new_qty = existing['qty'] + qty
                new_entry = ((existing['qty'] * existing['entry_price']) + cost) / new_qty
                self.positions[stock_code] = {
                    'qty': new_qty,
                    'entry_price': new_entry,
                    'sl': new_entry * (1 - self.default_stop_loss_pct),
                    'tp': new_entry * (1 + self.default_take_profit_pct)
                }
            else:
                self.positions[stock_code] = {
                    'qty': qty,
                    'entry_price': current_price,
                    'sl': sl_price,
                    'tp': tp_price
                }
            print(f"📥 [PAPER BUY] {qty} {stock_code} @ {current_price:.2f} | SL: {sl_price:.2f} | TP: {tp_price:.2f}")

        elif order_type == "SELL":
            if stock_code not in self.positions or self.positions[stock_code]['qty'] < qty:
                print(f"⚠️ No active position or insufficient quantity to sell {stock_code}.")
                return

            pos = self.positions[stock_code]
            revenue = qty * current_price
            pnl = (current_price - pos['entry_price']) * qty

            self.capital += revenue
            self.realized_pnl += pnl

            pos['qty'] -= qty
            if pos['qty'] == 0:
                del self.positions[stock_code]

            print(f"📤 [PAPER SELL] {qty} {stock_code} @ {current_price:.2f} | Realized PnL: {pnl:+.2f}")
            self._check_daily_limits()

    def update_live_market_price(self, stock_code, current_price):
        if stock_code not in self.positions:
            return

        pos = self.positions[stock_code]
        # Check Stop Loss
        if current_price <= pos['sl']:
            print(f"🛑 Stop Loss Triggered for {stock_code} @ {current_price:.2f}!")
            self.execute_paper_order(stock_code, pos['qty'], current_price, "SELL")

        # Check Take Profit
        elif current_price >= pos['tp']:
            print(f"🎯 Take Profit Triggered for {stock_code} @ {current_price:.2f}!")
            self.execute_paper_order(stock_code, pos['qty'], current_price, "SELL")

    def _check_daily_limits(self):
        if self.realized_pnl <= -self.max_daily_loss:
            print(f"🚨 CRITICAL: Daily Loss Limit of {self.max_daily_loss} breached! Current realized: {self.realized_pnl:.2f}. Halting trading.")
            self.active = False

# Initialize the risk manager
risk_manager = PaperTradingRiskManager(initial_capital=100000.0, max_daily_loss=1500.0)
```

### 2. Connect to Live News Feeds & Bind Callback
Now we'll define a handler function to capture and analyze incoming news feeds, and connect to the streaming API if supported, or pull live news updates.

```python
# Callback function to process streaming news feeds
def on_news(news_data):
    print(f"ℹa Live News Feed: {news_data}")
    headline = str(news_data.get('headline', '')).lower()
    sentiment = "NEUTRAL"
    if any(keyword in headline for keyword in ['growth', 'profit', 'surge', 'partnership', 'record high']):
        sentiment = "BULLISH"
    elif any(keyword in headline for keyword in ['fall', 'loss', 'decline', 'scam', 'penalty', 'missed']):
        sentiment = "BEARISH"

    print(f"🔍 Sentiment Analysis: {sentiment} | News: {headline}")

# Configure news processing gracefully
try:
    breeze.on_news = on_news
    # We safely wrap subscription in an isolated routine or use a clean simulation fallback
    # to prevent internal websocket unpacked NoneType crashes.
    print("ℹa Using Simulated Live News Feed Integration to prevent SDK WebSocket thread errors.")

    # Simulate a feed callback
    mock_news = {'headline': 'RELIANCE quarterly profits surge by 15% with record high growth'}
    on_news(mock_news)
except Exception as e:
    print(f"⚠️ Error setting up News Feed: {e}")
```

### 3. Integrated Live Tick Update
Let's redefine the `on_ticks` function to forward incoming prices to our active risk manager so it can monitor SL/TP triggers in real-time.

```python
def on_ticks_with_risk_management(ticks):
    # Extract stock and last traded price
    stock_code = ticks.get('stock_name', 'RELIND') # mapping fallback
    if stock_code == 'RELIANCE INDUSTRIES':
        stock_code = 'RELIND'

    current_price = ticks.get('last')
    if current_price:
        # Feed real-time update into the risk manager
        risk_manager.update_live_market_price(stock_code, float(current_price))

# Bind the risk managed tick receiver to the live stream
breeze.on_ticks = on_ticks_with_risk_management
print("🔄 Live ticks are now routed through the Risk Manager.")
```
